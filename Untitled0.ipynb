```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
# BAGIAN A - EKSPLORASI DATA AWAL
```

```python
## Load Dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
df['species_name'] = df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
df.head()
```

```python
print(df.median(numeric_only=True))
```

```python
## A2 Statistik Deskriptif
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
## A3a Histogram
print("Membuat Histogram...")
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
kolom_numerik = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']

for i, col in enumerate(kolom_numerik):
    sns.histplot(df[col], kde=True, ax=axes[i//2, i%2], color='skyblue')
    axes[i//2, i%2].set_title(f'Distribusi {col}')

plt.tight_layout()
plt.show()
```

Berdasarkan grafik histogram dan boxplot yang dihasilkan :


1. Bentuk Distribusi

Variabel sepal length dan sepal width cenderung memiliki distribusi yang mendekati normal (simetris). Sebaliknya, petal length dan petal width menunjukkan distribusi bimodal atau memiliki dua puncak, karena adanya perbedaan ukuran yang ekstrem antara spesies Setosa dan spesies lainnya.


2. Sebaran Data

Data pada variabel sepal memiliki sebaran yang lebih rapat atau terkonsentrasi. Sementara itu, variabel petal (terutama petal length) memiliki sebaran data yang sangat luas, mencakup rentang nilai dari yang sangat kecil hingga sangat besar.


```python
## A3b Boxplot
kolom_numerik = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']

plt.figure(figsize=(12, 10))

for i, col in enumerate(kolom_numerik, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df[col], color='lightgreen')
    plt.title(f'Boxplot {col}')
    plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
## BAGIAN B - BOOTSTRAP SAMPLING
```

```python
## Bootstrap Tunggal
sample_bootstrap = df['petal length (cm)'].sample(n=len(df), replace=True)
print(f"Mean Bootstrap: {sample_bootstrap.mean()}")
print(f"Mean Asli: {df['petal length (cm)'].mean()}")
```

```python
## B2 Bootstrap Berulang
def do_bootstrap(data, iterations):
    means = []
    for _ in range(iterations):
        sample = data.sample(n=len(data), replace=True)
        means.append(sample.mean())
    return means

means_100 = do_bootstrap(df['petal length (cm)'], 100)
means_1000 = do_bootstrap(df['petal length (cm)'], 1000)
means_10000 = do_bootstrap(df['petal length (cm)'], 10000)
```

```python
## B3 Histogram Bootstrap
data_target = df['petal length (cm)']

def bootstrap_means(data, iterations):
    means = []
    for _ in range(iterations):
        sample = data.sample(n=len(data), replace=True)
        means.append(sample.mean())
    return means

means_100 = bootstrap_means(data_target, 100)
means_1000 = bootstrap_means(data_target, 1000)
means_10000 = bootstrap_means(data_target, 10000)

plt.figure(figsize=(15, 5))

# Plot untuk 100 iterasi
plt.subplot(1, 3, 1)
sns.histplot(means_100, kde=True, color='skyblue')
plt.title('Bootstrap 100 Kali')
plt.xlabel('Mean Petal Length')

# Plot untuk 1000 iterasi
plt.subplot(1, 3, 2)
sns.histplot(means_1000, kde=True, color='pink')
plt.title('Bootstrap 1000 Kali')
plt.xlabel('Mean Petal Length')

# Plot untuk 10000 iterasi
plt.subplot(1, 3, 3)
sns.histplot(means_10000, kde=True, color='lightgreen')
plt.title('Bootstrap 10000 Kali')
plt.xlabel('Mean Petal Length')

plt.tight_layout()
plt.show()
```

Berdasarkan histogram distribusi mean bootstrap, terlihat bahwa hasil bootstrap untuk variabel petal length membentuk distribusi yang mendekati normal. Pada bootstrap sebanyak 100 kali, distribusi mean masih terlihat cukup menyebar dan belum stabil. Hal ini menunjukkan bahwa jumlah pengulangan yang sedikit masih menghasilkan variasi estimasi yang cukup besar.

Pada bootstrap sebanyak 1000 kali, distribusi mean mulai terlihat lebih halus dan terpusat di sekitar nilai rata-rata data asli. Variasi hasil bootstrap juga menjadi lebih kecil dibandingkan dengan bootstrap 100. Sementara itu, pada bootstrap sebanyak 10000 kali, distribusi mean terlihat paling stabil, lebih rapat, dan semakin mendekati bentuk distribusi normal.


```python
## B3b Boxplot Mean Bootstrap
bootstrap_data = pd.DataFrame({
    'Iterasi': ['100'] * 100 + ['1000'] * 1000 + ['10000'] * 10000,
    'Mean': means_100 + means_1000 + means_10000
})

plt.figure(figsize=(10, 6))
sns.boxplot(x='Iterasi', y='Mean', data=bootstrap_data, palette='Set3')

plt.title('Perbandingan Boxplot Mean Bootstrap (100, 1000, 10000 Iterasi)')
plt.xlabel('Jumlah Iterasi Bootstrap')
plt.ylabel('Rata-rata Petal Length (cm)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

Hasil boxplot juga menunjukkan bahwa rentang nilai mean bootstrap semakin sempit seiring bertambahnya jumlah bootstrap. Bootstrap 100 memiliki penyebaran yang paling lebar, sedangkan bootstrap 10000 memiliki penyebaran paling kecil. Hal ini menunjukkan bahwa semakin banyak jumlah bootstrap, estimasi rata-rata menjadi semakin stabil dan konsisten.

```python
## B4 Confidance Interval 95%
import numpy as np

ci_100 = np.percentile(means_100, [2.5, 97.5])
ci_1000 = np.percentile(means_1000, [2.5, 97.5])
ci_10000 = np.percentile(means_10000, [2.5, 97.5])

print("=== INTERVAL KEPERCAYAAN 95% (95% CI) ===")
print(f"Bootstrap 100 kali   : {ci_100[0]:.4f} sampai {ci_100[1]:.4f}")
print(f"Bootstrap 1000 kali  : {ci_1000[0]:.4f} sampai {ci_1000[1]:.4f}")
print(f"Bootstrap 10000 kali : {ci_10000[0]:.4f} sampai {ci_10000[1]:.4f}")

mean_asli = df['petal length (cm)'].mean()
print(f"\nMean Data Asli: {mean_asli:.4f}")
```
